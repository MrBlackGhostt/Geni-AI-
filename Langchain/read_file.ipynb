```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.vectorstores import InMemoryVectorStore
```

```python
loader = WebBaseLoader(
    web_paths=("https://vkoskiv.com/first-linux-patch/",),

    
)
docs = loader.load()
text_spillter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunk = text_spillter.split_documents(docs)
```

```python
print(chunk)
```

```python
from langchain_openai import OpenAIEmbeddings
api_key=""
embeddings =  OpenAIEmbeddings(model="text-embedding-3-large", api_key="")
```

```python
vector_store = InMemoryVectorStore(embeddings)
vector_store.add_documents(chunk)
```

```python
print(vector_store)
```

```python
retrieved_docs = vector_store.similarity_search("how did he do the first contribute")
```
